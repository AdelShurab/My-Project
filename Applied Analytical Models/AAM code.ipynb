Data Reduction 

```python
#by use sampling to reduce the size of a dataset
# Define the sampling fraction (0.5 for 50% reduction)
#a value of 0.5 will retain 50% of the rows in the dataset
sample_fraction = 0.5

# Perform random sampling to reduce the dataset size
dataset_reduced = dataset.sample(frac=sample_fraction)
```

```python
#put the result of the reduction in a new csv file 
dataset_reduced.to_csv('reduced_data.csv', index=False)
```
