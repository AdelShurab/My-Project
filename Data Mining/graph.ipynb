```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
graph = nx.Graph()
```

```python
graph.add_node("A")
graph.add_node("B")
graph.add_node("C")
graph.add_node("D")
graph.add_node("E")
```

```python
graph.add_edge("A", "B")
graph.add_edge("B", "C")
graph.add_edge("C", "D")
graph.add_edge("D", "E")
graph.add_edge("E", "A")
```

```python
nx.draw(graph, with_labels=True)
plt.show()
```

```python
clustering = nx.clustering(graph)
```

```python
pos = nx.spring_layout(graph)
nx.draw(graph, pos, with_labels=True)
nx.draw_networkx_labels(graph, pos, labels=clustering)
plt.show()
```
